```python
#test path (print number of cs files inside, recursivly checking subfolders)
import os

def count_cs_files(path):
    count = 0
    for root, dirs, files in os.walk(path):
        count += sum(1 for file in files if file.endswith('.cs'))
    return count

num_cs_files = count_cs_files(SCRIPTS_PATH)
print(f"Number of .cs files in '{SCRIPTS_PATH}': {num_cs_files}")
```

```python
#squash it's content to the one string. Start every file with --- filename.cs ---

def squash_cs_files(path):
    content = ""
    for root, dirs, files in os.walk(path):
        for file in files:
            if file.endswith('.cs'):
                with open(os.path.join(root, file), 'r', encoding='utf-8-sig') as f:
                    file_content = f.read()
                    content += f"--- {file} ---\n{file_content}\n"
    return content

squashed_content = squash_cs_files(SCRIPTS_PATH)
```

```python
#save squashed content to a file
with open("squashed_scripts.cs", "w", encoding="utf-8") as f:
    f.write(squashed_content)
```

```python
# restore files from squashed content 
# create a dictionary filename.cs to filecontent

def restore_files_from_squashed(content):
    files = {}
    current_file = None
    for line in content.splitlines():
        if line.startswith("--- ") and line.endswith(" ---"):
            current_file = line[4:-4]
            files[current_file] = ""
        elif current_file is not None:
            files[current_file] += line + "\n"
    return files
```

```python
RESTORE_PATH = "./squashed_scripts_refactored.cs"
```

```python
#load the file from RESTORE_PATH
with open(RESTORE_PATH, "r",encoding="utf-8-sig") as f:
    restored_content = f.read()

restored_files = restore_files_from_squashed(restored_content)
```

```python
#apply all the changes: for each key in restored files, find the original file within SCRIPTS_PATH and replace its content with the value from restored files
for filename, content in restored_files.items():
    # Get original filepath by going through all files in the Scripts_Path and checking if the file name matches
    original_file_path = None
    for root, dirs, files in os.walk(SCRIPTS_PATH):
        if filename in files:
            original_file_path = os.path.join(root, filename)
            break

    if original_file_path:
        with open(original_file_path, "w", encoding="utf-8-sig") as f:
            f.write(content)
    else:
        print(f"Original file for {filename} not found. Skipping.")
```
